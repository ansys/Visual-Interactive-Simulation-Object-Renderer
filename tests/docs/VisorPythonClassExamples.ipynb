```python
from pathlib import Path
```

```python
from ansys.visor.viewer import Visor
```

```python
DATASET1_PATH = "../../examples/assets/vtk_scene_sphere_l2_b3_r32_v3_c1_z0.vtm"
DATASET1_METADATA_PATH = "../../examples/assets/vtk_scene_sphere_l2_b3_r32_v3_c1_z0.json"

DATASET2_PATH = "../../examples/assets/vtk_scene_sphere_l2_b3_r32_v3_c1_z3.vtm"
DATASET2_METADATA_PATH = "../../examples/assets/vtk_scene_sphere_l2_b3_r32_v3_c1_z3.json"
```

```python
####################################################################
# Step 0: Instantiate a Visor class
####################################################################
visualizer = Visor() # spins up Visor class on URL http://localhost:8081
```

```python
####################################################################
# Step 1: Start a Visor session with multiblock + metadata
####################################################################

visualizer.start(input=DATASET1_PATH, metadata=DATASET1_METADATA_PATH)
```

```python
# Check that the dataset loads correctly:
#    * meshes look as expected
#    * the opacities are correct in the slider
#    * and opacities are applied correctly on the mesh.
```

```python
####################################################################
# Step 2: Print info about the instance
####################################################################
visualizer.info()
```

```python
####################################################################
# Step 3: Adjust the settings on parts
# 
# Adjust settings on the following parts:
# * level_0_block_0_part_0_parent:
#    * Adjust opacity to something high but not 100%, e.g. 95%
#    * Color by constant (i.e. don't change the dropdown color by variable value)
#    * Change the Constant Color to something different
# * level_0_block_0_part_1_parent:
#    * Color by POINT -random_variable_2 (Scalar)
# * level_0_block_0_part_2_parent:
#    * Color by POINT - gradient_variable_0 (Scalar)
# * level_1_block_1_part_1 (or any individual sphere):
#    * Toggle visibility OFF
#    * Check that all settings are applied correctly.
####################################################################

```

```python
####################################################################
# Step 4: Add a new dataset to the scene
####################################################################
visualizer.add_dataset(input=DATASET2_PATH, metadata=DATASET2_METADATA_PATH)
```

```python
# Check that the new dataset loads correctly:
#    * meshes look as expected
#    * the opacities are correct in the slider
#    * and opacities are applied correctly on the mesh.
# Check that the first dataset is still present, with part properties preserved from before the add_dataset operation.
```

```python
####################################################################
# Step 5. List datasets
####################################################################
visualizer.list_datasets()
```

```python
####################################################################
# Step 6. Remove the second dataset
####################################################################
visualizer.remove_dataset(2731261597931916)
```

```python
# Check that
#   * the dataset was removed in the viewer, and
#   * the other dataset still has the expected per-part settings applied.
```

```python
####################################################################
# Step 7. Save the state
####################################################################
await visualizer.save_state("test_dir")
```

```python
# Check that the state visor.json file and dataset vtkhdf file were serialized to disk
for item in Path("test_dir").iterdir():
    print(item.name)
```

```python
####################################################################
# Step 7. Load the state on already-loaded dataset
####################################################################

# Before loading the state, change scene settings (eg toggle visibility,
# color-by, or expand/minimize some panels, or change the camera
visualizer.load_state("test_dir")
```
