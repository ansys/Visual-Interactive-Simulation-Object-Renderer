```python
from IPython.display import IFrame

from ansys.visor.viewer import Visor

asset_path1 = '../../examples/assets/vtk_scene_sphere_l2_b3_r32_v3_c1_z0.vtm'
asset_path2 = '../../examples/assets/tensors9.vtp'

#######################################################
# Example running Visor in a jupyter notebook
# Shows how to create two separate Visor visualizers
# on two distinct ports (8081, 8082) and
# view them in an iframe
#######################################################
```

```python
# Instantiate a Visor visualizer on port 8081
visor = Visor(url='http://127.0.0.1:8081')
```

```python
# Start the visualizer using the first asset
dataset_id1 = visor.start(asset_path1)
```

```python
# NBVAL_IGNORE_OUTPUT
# View the visualizer inline
IFrame(src='http://127.0.0.1:8081', width=1200, height=600)
```

```python
dataset_id2 = visor.update(asset_path2)
```

```python
# Print the information about the first visualizer instance
visor.info()
```

```python
# NBVAL_IGNORE_OUTPUT
# Initialize a second Visor visualizer on port 8082
visor1 = Visor(url='http://127.0.0.1:8082')
```

```python
dataset_id3 = visor1.start(asset_path2)
```

```python
# NBVAL_IGNORE_OUTPUT
IFrame(src='http://127.0.0.1:8082', width=1200, height=600)
```

```python
import requests

r = requests.get("http://localhost:8081/index.html", timeout=5)
print(r.status_code, r.headers.get("content-type"))
r = requests.get("http://localhost:8082/index.html", timeout=5)
print(r.status_code, r.headers.get("content-type"))
```

```python
visor.stop()
```

```python
visor1.stop()
```
