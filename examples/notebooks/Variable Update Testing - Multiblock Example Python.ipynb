```python

import numpy as np
from IPython.display import IFrame
```

```python
from ansys.visor.viewer import Visor
```

```python
############################
# Initialize and start Visor
############################
```

```python
vis = Visor()
```

```python
dataset_id1 = vis.start("../../examples/assets/vtk_scene_sphere_l2_b3_r32_v3_c1_z0.vtm", metadata="../../examples/assets/vtk_scene_sphere_l2_b3_r32_v3_c1_z0.json")
```

```python
IFrame("http://localhost:8081", width="1000", height="500")
```

```python
# Print dataset metadata and get the dataset ID
datasets = vis.list_datasets()
# print(json.dumps(datasets, indent=4))
# Print the dataset ID
print(f"dataset ID: {dataset_id1}")
```

```python
parts = vis.list_variables(dataset_id1)
```

```python
# Create the new variables for the test point dataArray (zeros):
part = parts[0]
part.to_dict()
```

```python
variables = part.variables
var_names = [f"{v.name}::{v.type.value}::{v.num_components}" for v in variables]
var_to_update = variables[0]

print(f"{part.part_name} has the following variables:")
print('   ' + '\n   '.join(var_names))
print(f"variable to update: {var_to_update.name}::{var_to_update.type.value}::{var_to_update.num_components}")

```

```python
# Create new vector data
new_vector_values = np.zeros((var_to_update.num_points, var_to_update.num_components))
```

```python
# Compile a dict with the vector variable metadtata + updated values
vector_update_info = {
    "type": "point",
    "name": var_to_update.name,
    "num_components": var_to_update.num_components,
    "data": new_vector_values,
    "part_id": part.part_id
}
```

```python
# Run the actual update command
vis.update_variables(dataset_id1, [vector_update_info])
```
