```python
from IPython.display import IFrame

from ansys.visor.viewer import Visor
```

```python
# Initialize Visor on port 8083
visualizer = Visor(url='http://localhost:8083')
```

```python
# Provide initial input via the `start` method.  
# Note: the input and metadata are optional.  
# If we just run start() with no args before adding datasets, it will render a blank scene.
dataset_id1 = visualizer.start('../../examples/assets/vtk_scene_sphere_l2_b3_r32_v3_c1_z0.vtm', "../../examples/assets/vtk_scene_sphere_l2_b3_r32_v3_c1_z0.json")
```

```python
# View the visualizer inline
IFrame(src='http://localhost:8083', width=1000, height=600)
```

```python
# Add a second dataset
dataset_id2 = visualizer.add_dataset('../../tests/files/plate.vtp')
```

```python
datasets = visualizer.list_datasets()
dataset_ids = list(datasets)
datasets
```

```python
visualizer.remove_dataset(dataset_id1)
```

```python
visualizer.remove_dataset(dataset_id2)
```
