```python

import numpy as np
from IPython.display import IFrame
```

```python
from ansys.visor.viewer import Metadata, Visor
```

```python
############################
# Initialize and start Visor
############################
```

```python
vis = Visor()
```

```python
dataset_id1 = vis.start("../../tests/files/plate.vtp", metadata=Metadata(name="sphere_1", unit="m"))
```

```python
IFrame("http://localhost:8081", width="1000", height="500")
```

```python
# Print dataset metadata and get the dataset ID
datasets = vis.list_datasets()
# print(json.dumps(datasets, indent=4))
# Print the dataset ID
print(f"dataset ID: {dataset_id1}")
```

```python
parts = vis.list_variables(dataset_id1)
part_to_update = parts[0]
part_to_update
```

```python
# Create the new variables for the test point dataArray (zeros):
variable_to_update = part_to_update.variables[0]
new_vector_values = np.random.rand(variable_to_update.num_points, variable_to_update.num_components)
```

```python
# Compile a dict with the vector variable metadtata + updated values
vector_update_info = {
    "type": "point",
    "name": variable_to_update.name,
    "num_components": variable_to_update.num_components,
    "data": new_vector_values
}
```

```python
# Run the actual update command
vis.update_variables(dataset_id1, [vector_update_info])
```
